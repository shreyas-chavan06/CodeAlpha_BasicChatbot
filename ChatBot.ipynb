```python
# Function to run chatbot
def chatBot():
    # Keep the conversation running till the user says bye.
    while True:
        # Takes input from a user and turn it to lowercase to handle capitalizations
        user_input=input("You: ").lower()
        # Responds when the user sends Hello
        if user_input == "hello":
            print("ChatBot: - Hii, Welcome to ChatBot")
        # Responds when the user asks how the chatbot is
        elif user_input=="how are you":
            print("ChatBot: - I am fine, thanks!")
        # Responds when the user says bye and the conversation stops
        elif user_input=="bye":
            print("ChatBot: - GoodBye..")
            break
        # Responds to unknown message
        else:
            print("ChatBot: - Sorry I didn't Understand that")
# Start the chatbot
chatBot()
```
